#Dictionary -{}
{ key :value,
key2 :value,
key3:value
}
# properties of dict
1. dict is mutable
2. the dict key should be unique,values can have duplicates
3. values can be have  any datatypes(int,float,str,list,tuple,set,bool,dict)
   # Dict methods
# methods                # purpose
1. keys()                Get all keys
2. values                Get all values
3. items()               Get key-value pair
4. get()                 Get a value safely
5. update()              Add/update multiple items
6. pop()                 Remove an item
7. popitem()             Remove the last item
8. clear()               Remove every thing
9. copy()                Create a copy 

```python
dict_a={
    "cse":3,
    "ece":4,
    "eee":5
}
print(dict_a)
print(dict_a.keys())
print(dict_a.values())
```

```python
dict_a={
    "cse":3,
    "ece":4,
    "eee":5
}
dict_a['ece']=5     #update particular key
print(dict_a)
```

```python
dict_a={
    "cse":3,
    "ece":4,
    "eee":5
}
dict_a.update({'aiml':2})
print(dict_a)
```

```python
dict_a={
    "cse":3,
    "ece":4,
    "eee":5
}
dict_a.update({'aiml':2,'ds':4}) # update the no.of keys to seperate with ,
print(dict_a)
```

```python
dict_a={
    "cse":3,
    "ece":4,
    "eee":5
}
for key,value in dict_a.items():
    print(key,":",value)
```

```python
dict_a={
    "cse":3,
    "ece":4,
    "eee":5
}
for key in dict_a.keys():
    print(key)
```

```python
for v in dict_a.values():
    print(v)
```
